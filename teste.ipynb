# Visualizando arquivo Parquet
Para ler e visualizar arquivos `.parquet` em Python usando Pandas, precisamos instalar a biblioteca `pandas` e uma engine para ler arquivos parquet, como `fastparquet` ou `pyarrow`.

Você pode instalar as dependências no terminal com:
```bash
pip install pandas pyarrow
```

````python
import pandas as pd

# Substitua 'caminho_do_arquivo.parquet' pelo caminho real do seu arquivo
caminho_arquivo = 'merged/producao_Mar.parquet'

try:
    # Lendo o arquivo parquet
    df = pd.read_parquet(caminho_arquivo)
    
    # Exibindo as primeiras linhas do DataFrame
    display(df.head())
    
    # Exibindo informações sobre o DataFrame (colunas, tipos de dados, uso de memória)
    print("\nInformações do arquivo:")
    display(df.info())
    
except Exception as e:
    print(f"Ocorreu um erro ao tentar ler o arquivo: {e}")
````
